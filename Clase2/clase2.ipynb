## listas y diccionarios
> listas:
coleccion ordenada, implementada como un array dinamico, su acceso lo tiene por indice.



```python
# lista de estudiantes

estudiantes = ["graciela","andres","carolina"]

print(estudiantes[1])
# -1 para acceder al ultimo de la lista
print(estudiantes[-1])
#decir el largo de la lista
print(len(estudiantes))
#agregar un estudiante a la lista
estudiantes.append("luis")
print(estudiantes)
```

# diccionarios

pares de clave - valor

```python
# registro de estudiantes
estudiantes = {
    "nombre": "yamel calderon",
    "edad": 45,
    "correo": "yamell2010@gmail.com",
    "activo": True,
    "promedio": 8.9
}

print(estudiantes["nombre"])
print(f"estudiante: {estudiantes['nombre']}")
print(f"edad: {estudiantes['edad']}")
print(f"correo: {estudiantes['correo']}")
print(f"activo: {estudiantes['activo']}")
print(f"promedio: {estudiantes['promedio']}")

# actualizo un valr del diccionario

estudiantes["activo"] = False
print(f"{estudiantes["activo"]}")
```

```python
# se puede hacer una lista de diccionarios
estudiante = {
    "nombre": "yamel calderon",
    "edad": 45,
    "correo": "yamell2010@gmail.com",
    "activo": True,
    "promedio": 8.9
}

estudiantes = [
    estudiante,
    {
    "nombre": "pedro calderon",
    "edad": 45,
    "correo": "pedro2010@gmail.com",
    "activo": True,
    "promedio": 8.9
    },
    {
    "nombre": "luis calderon",
    "edad": 45,
    "correo": "luis2010@gmail.com",
    "activo": True,
    "promedio": 8.9
    }
]

print(estudiantes)
```

# condicionales y bucles 

> condicionales if/el if/else, es una estructura de control, controla el flujo del programa 


```python
nombre = "Ana"
promedio = 8.7

if promedio >= 7:
    print(f"felicitaciones {nombre}")
else:
    print(f"necesitas mejorar {nombre}")
```

```python
nombre = "pepito"
promedio = 5

if promedio >= 7:
    print(f"felicitaciones {nombre}")
elif promedio >= 5:
    print(f"{nombre} puedes mejorar")
else:
    print(f"necesitas mejorar {nombre}")
```

## bucle
> for, es una estructura de iteracion, recorre todo el elemento


```python
learners = [
    {"nombre": "pepita", "promedio": 8.7},
    {"nombre": "pepito", "promedio": 7.5},
    {"nombre": "andres", "promedio": 9.1},
]

for learner in learners:
    if learner["promedio"] >= 7:
        print(f"felicitaciones {learner["nombre"]}")
    elif learner["promedio"] >= 5:
        print(f"{nombre} puedes mejorar")
    else:
        print(f"necesitas mejorar {nombre}")
```
