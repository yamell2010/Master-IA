## 1. Variables y tipos de datos

Una **variable** es un nombre que apunta a un valor guardado en memoria. Permite trabajar con información dinámica a lo largo del programa.

Los **f-strings** (`f"..."`) permiten insertar el valor de una variable directamente dentro de un texto usando `{}`.

### Los 4 tipos esenciales

| Tipo | Ejemplo | Uso en inventario | Descripción |
|------|---------|-------------------|-------------|
| `str` | `"LAP-001"` | nombre, código, categoría | Texto entre comillas |
| `int` | `14` | stock, unidades | Número entero sin decimales |
| `float` | `2350000.0` | precio, descuento | Número con parte decimal |
| `bool` | `True / False` | disponible, en_oferta | Solo dos valores posibles |

````python
# Registro de un producto
nombre_producto = "Laptop Lenovo IdeaPad"   # str
codigo          = "LAP-001"                  # str
precio          = 2350000.0                  # float
stock           = 14                         # int
disponible      = True                       # bool

# Imprimir con f-strings
print(f"Producto:   {nombre_producto}")
print(f"Código:     {codigo}       → tipo: {type(codigo).__name__}")
print(f"Precio:     ${precio:,.0f}   → tipo: {type(precio).__name__}")
print(f"Stock:      {stock} unidades  → tipo: {type(stock).__name__}")
print(f"Disponible: {disponible}        → tipo: {type(disponible).__name__}")
````

````python
# Operaciones con variables numéricas
precio          = 2350000.0
descuento       = 0.15
unidades_mes    = 8

precio_final    = precio * (1 - descuento)
ingreso_mensual = precio_final * unidades_mes

print(f"Precio original:    ${precio:,.0f}")
print(f"Precio con 15% dto: ${precio_final:,.0f}")
print(f"Ingreso del mes:    ${ingreso_mensual:,.0f}")
````

## 2. Listas
Una **lista** es una colección ordenada y mutable de elementos. Se define con corchetes `[]` y se accede a cada elemento por su índice (comenzando en `0`).

```python
mi_lista = [elemento0, elemento1, elemento2]
#                 ↑           ↑           ↑
#            índice 0    índice 1    índice 2
```

````python
# Crear y acceder a elementos
productos = ["Laptop Lenovo", "Mouse Logitech", "Teclado mecánico", "Monitor Samsung"]

print(productos[0])    # Primer elemento
print(productos[-1])   # Último elemento (índice negativo)
print(productos[1:3])  # Slice: del índice 1 al 2 (el 3 no se incluye)
print(len(productos))  # Cantidad total de elementos
````

### Métodos principales de listas

| Método | Qué hace |
|--------|----------|
| `.append(x)` | Agrega `x` al final |
| `.insert(i, x)` | Inserta `x` en la posición `i` |
| `.extend([...])` | Agrega todos los elementos de otra lista al final |
| `.remove(x)` | Elimina la primera aparición de `x` |
| `.pop(i)` | Elimina y retorna el elemento en posición `i` (por defecto el último) |
| `.index(x)` | Retorna el índice de la primera aparición de `x` |
| `.count(x)` | Cuenta cuántas veces aparece `x` |
| `.sort()` | Ordena la lista en su lugar (modifica la original) |
| `.reverse()` | Invierte la lista en su lugar |
| `.copy()` | Retorna una copia independiente |
| `.clear()` | Vacía la lista completamente |
| `sorted(lista)` | Retorna una nueva lista ordenada sin modificar la original |
| `x in lista` | Verifica si `x` existe en la lista (`True` / `False`) |

````python
# ── append / insert / extend ─────────────────────────────────────
catalogo = ["Laptop", "Mouse", "Teclado"]

catalogo.append("Monitor")                     # Agrega al final
catalogo.insert(1, "Webcam")                    # Inserta en posición 1
catalogo.extend(["Auriculares", "Mousepad"])    # Agrega varios al final

print(catalogo)
````

````python
# ── remove / pop ─────────────────────────────────────────────────
catalogo = ["Laptop", "Webcam", "Mouse", "Teclado", "Monitor"]

catalogo.remove("Webcam")         # Elimina por valor
print("Después de remove:", catalogo)

ultimo = catalogo.pop()           # Elimina y retorna el último
print(f"Elemento extraído con pop(): {ultimo}")
print("Lista resultante:", catalogo)

segundo = catalogo.pop(1)         # Elimina y retorna el de posición 1
print(f"pop(1) extrajo: {segundo}")
print("Lista resultante:", catalogo)
````

````python
# ── sort / sorted / reverse ──────────────────────────────────────
precios = [950000, 85000, 2350000, 320000, 780000]

print("Original:",         precios)
print("sorted() (nuevo):", sorted(precios))              # No modifica la original
print("Original sigue:",   precios)                      # Comprobación

precios.sort()                                            # Modifica en su lugar
print("sort() de menor a mayor:", precios)

precios.sort(reverse=True)                               # Orden descendente
print("sort(reverse=True):", precios)

precios.reverse()                                        # Solo invierte
print("reverse():", precios)
````

````python
# ── index / count / in ───────────────────────────────────────────
categorias = ["electrónica", "periféricos", "electrónica", "mobiliario", "electrónica"]

print(f"Posición de 'mobiliario':          {categorias.index('mobiliario')}")
print(f"Veces que aparece 'electrónica':   {categorias.count('electrónica')}")
print(f"¿Existe 'papelería'?:              {'papelería' in categorias}")
print(f"¿Existe 'periféricos'?:            {'periféricos' in categorias}")
````

````python
# ── copy / clear ─────────────────────────────────────────────────
original  = ["Laptop", "Mouse", "Teclado"]
copia     = original.copy()     # Copia independiente

copia.append("Monitor")         # No afecta a la original
print("Original:", original)
print("Copia:   ", copia)

copia.clear()                   # Vacía la copia
print("Copia después de clear():", copia)
````

````python
# ── Funciones útiles sobre listas numéricas ──────────────────────
precios = [2350000, 85000, 320000, 950000, 780000]

print(f"Total productos: {len(precios)}")
print(f"Precio mínimo:   ${min(precios):,}")
print(f"Precio máximo:   ${max(precios):,}")
print(f"Suma total:      ${sum(precios):,}")
print(f"Precio promedio: ${sum(precios)/len(precios):,.0f}")
````

## 3. Diccionarios

Un **diccionario** almacena pares `clave: valor`. Es ideal para representar un objeto con múltiples atributos — como un producto con nombre, precio, stock, etc.

```python
mi_dict = {
    "clave1": valor1,
    "clave2": valor2
}
```

````python
# Un producto como diccionario
producto = {
    "nombre":    "Laptop Lenovo IdeaPad",
    "codigo":    "LAP-001",
    "precio":    2350000.0,
    "stock":     14,
    "categoria": "electrónica",
    "en_oferta": False
}

# Acceder por clave
print(producto["nombre"])
print(producto["precio"])

# .get() — acceso seguro (no lanza error si la clave no existe)
print(producto.get("descuento", "Sin descuento"))   # Retorna valor por defecto
````

### Métodos principales de diccionarios

| Método / Operación | Qué hace |
|---|---|
| `d["clave"]` | Accede al valor (lanza error si no existe) |
| `d.get("clave", default)` | Accede al valor sin error; retorna `default` si no existe |
| `d["clave"] = valor` | Agrega o actualiza un campo |
| `d.update({...})` | Actualiza múltiples campos a la vez |
| `del d["clave"]` | Elimina un campo |
| `d.pop("clave")` | Elimina y retorna el valor del campo |
| `d.setdefault("clave", val)` | Agrega el campo con `val` solo si no existe |
| `d.keys()` | Retorna todas las claves |
| `d.values()` | Retorna todos los valores |
| `d.items()` | Retorna pares `(clave, valor)` — ideal para `for` |
| `"clave" in d` | Verifica si una clave existe |
| `d.copy()` | Retorna una copia independiente |
| `len(d)` | Número de pares clave-valor |

````python
# ── Agregar / actualizar con [] y update() ───────────────────────
producto = {"nombre": "Laptop Lenovo", "precio": 2350000.0, "stock": 14}

producto["categoria"] = "electrónica"                          # Nuevo campo
producto["stock"]     = 10                                     # Actualizar campo
print("Después de agregar/actualizar:", producto)

producto.update({"en_oferta": True, "descuento": 0.15})       # Varios a la vez
print("Después de update():", producto)
````

````python
# ── del / pop() ──────────────────────────────────────────────────
producto = {
    "nombre":    "Laptop Lenovo",
    "precio":    2350000.0,
    "stock":     10,
    "descuento": 0.15,
    "temporal":  "borrar"
}

del producto["temporal"]                   # Eliminar con del (no retorna nada)
desc = producto.pop("descuento")           # Eliminar y guardar el valor
print(f"Descuento que fue eliminado: {desc}")
print("Producto resultante:", producto)
````

````python
# ── setdefault() — agregar solo si no existe ──────────────────────
producto = {"nombre": "Mouse Logitech", "precio": 85000, "stock": 40}

producto.setdefault("categoria", "sin categoría")    # Se agrega porque no existe
producto.setdefault("precio", 999999)                # NO se modifica: ya existe

print(producto)
````

````python
# ── keys() / values() / items() ──────────────────────────────────
producto = {
    "nombre":    "Laptop Lenovo IdeaPad",
    "precio":    2350000.0,
    "stock":     10,
    "categoria": "electrónica"
}

print("Claves:  ", list(producto.keys()))
print("Valores: ", list(producto.values()))
print()

# items() — recorrer con for
print("--- Ficha del producto ---")
for clave, valor in producto.items():
    print(f"  {clave:12s}: {valor}")
````

````python
# ── Verificar si una clave existe ────────────────────────────────
producto = {"nombre": "Monitor Samsung", "precio": 950000, "stock": 8}

if "descuento" in producto:
    print(f"Tiene descuento: {producto['descuento'] * 100}%")
else:
    print("Sin descuento registrado.")

# .get() como forma compacta de manejar claves opcionales
descuento    = producto.get("descuento", 0)
precio_final = producto["precio"] * (1 - descuento)
print(f"Precio final: ${precio_final:,.0f}")
````

````python
# ── copy() y len() ───────────────────────────────────────────────
producto    = {"nombre": "Teclado mecánico", "precio": 320000, "stock": 3}
respaldo    = producto.copy()      # Copia independiente

producto["stock"] = 0              # Modifica el original
print("Original modificado:", producto)
print("Respaldo sin cambios:", respaldo)
print(f"Número de campos: {len(producto)}")
````

````python
# ── Lista de diccionarios — estructura de un dataset real ────────
inventario = [
    {"nombre": "Laptop Lenovo",    "precio": 2350000, "stock": 14, "categoria": "electrónica"},
    {"nombre": "Mouse Logitech",   "precio":   85000, "stock": 40, "categoria": "periféricos"},
    {"nombre": "Teclado mecánico", "precio":  320000, "stock":  3, "categoria": "periféricos"},
    {"nombre": "Monitor Samsung",  "precio":  950000, "stock":  8, "categoria": "electrónica"},
    {"nombre": "Silla ergonómica", "precio":  780000, "stock":  0, "categoria": "mobiliario"},
]

# Acceder a un producto específico
print(inventario[2]["nombre"])    # Teclado mecánico

# Recorrer toda la lista
print("\n--- Catálogo completo ---")
for p in inventario:
    print(f"  {p['nombre']:25s}  ${p['precio']:>12,}   stock: {p['stock']}")
````

## 4. Condicionales

## 5. Bucles

Los bucles permiten ejecutar un bloque de código repetidamente. `for` se usa cuando la cantidad de repeticiones es conocida; `while` cuando se repite hasta que una condición deje de cumplirse.

````python
# for — recorrer el inventario
inventario = [
    {"nombre": "Laptop Lenovo",    "precio": 2350000, "stock": 14, "categoria": "electrónica"},
    {"nombre": "Mouse Logitech",   "precio":   85000, "stock": 40, "categoria": "periféricos"},
    {"nombre": "Teclado mecánico", "precio":  320000, "stock":  3, "categoria": "periféricos"},
    {"nombre": "Monitor Samsung",  "precio":  950000, "stock":  8, "categoria": "electrónica"},
    {"nombre": "Silla ergonómica", "precio":  780000, "stock":  0, "categoria": "mobiliario"},
]

print("Estado del inventario:")
for producto in inventario:
    s = producto["stock"]
    if   s == 0: alerta = "AGOTADO"
    elif s <= 5: alerta = "Crítico"
    else:        alerta = "OK"
    print(f"  {producto['nombre']:25s}  stock: {s:3d}  {alerta}")
````

````python
# enumerate() — cuando necesitas el índice
print("Nro  Producto")
for i, producto in enumerate(inventario, start=1):
    print(f"  {i}.  {producto['nombre']}")
````

````python
# Acumular con bucle
valor_total = 0
for producto in inventario:
    valor_total += producto["precio"] * producto["stock"]

print(f"Valor total del inventario: ${valor_total:,.0f}")
````

````python
# while — reabastecimiento automático
stock_teclado = 3
pedidos       = 0

while stock_teclado < 10:
    stock_teclado += 5
    pedidos       += 1
    print(f"  Pedido #{pedidos} → stock ahora: {stock_teclado}")

print(f"Stock normalizado en {pedidos} pedido(s).")
````

## 6. Funciones

Una **función** es un bloque de código reutilizable que recibe parámetros y puede retornar un resultado. Permite evitar la repetición y organizar el programa en piezas independientes (principio DRY: *Don't Repeat Yourself*).

```python
def nombre_funcion(parametro1, parametro2="valor_por_defecto"):
    """Descripción de lo que hace la función."""
    # lógica
    return resultado
```

````python
# Función básica con return
def clasificar_stock(stock):
    """Retorna el estado del stock."""
    if stock == 0:    return "Agotado"
    elif stock <= 5:  return "Stock crítico"
    elif stock <= 15: return "Stock normal"
    return "Stock alto"

# Reutilizable con cualquier valor
for valor in [0, 3, 14, 50]:
    print(f"  stock={valor:2d}  →  {clasificar_stock(valor)}")
````

````python
# Parámetro con valor por defecto
def precio_con_descuento(precio, descuento=0.10):
    """Retorna el precio final aplicando un descuento (10% por defecto)."""
    return precio * (1 - descuento)

print(f"Descuento por defecto (10%): ${precio_con_descuento(2350000):,.0f}")
print(f"Con 20% de descuento:        ${precio_con_descuento(2350000, 0.20):,.0f}")
print(f"Sin descuento (0%):          ${precio_con_descuento(2350000, 0):,.0f}")
````

````python
# Función que retorna múltiples valores
def resumen_categoria(inventario, categoria):
    """Retorna (total_productos, total_stock, valor_total) para una categoría."""
    filtro      = [p for p in inventario if p["categoria"] == categoria]
    total_stock = sum(p["stock"]               for p in filtro)
    valor_total = sum(p["precio"] * p["stock"]  for p in filtro)
    return len(filtro), total_stock, valor_total

inventario = [
    {"nombre": "Laptop Lenovo",    "precio": 2350000, "stock": 14, "categoria": "electrónica"},
    {"nombre": "Mouse Logitech",   "precio":   85000, "stock": 40, "categoria": "periféricos"},
    {"nombre": "Teclado mecánico", "precio":  320000, "stock":  3, "categoria": "periféricos"},
    {"nombre": "Monitor Samsung",  "precio":  950000, "stock":  8, "categoria": "electrónica"},
    {"nombre": "Silla ergonómica", "precio":  780000, "stock":  0, "categoria": "mobiliario"},
]

for cat in ["electrónica", "periféricos", "mobiliario"]:
    cant, stock_t, valor = resumen_categoria(inventario, cat)
    print(f"{cat:15s} → {cant} productos | {stock_t} uds | ${valor:,.0f}")
````
