```python
def prdecir_cliente(dias_de_compra, num_compra, gasto_total):
    if dias_de_compra < 30 and num_compra > 3:
        return "definiticamente piensa volver"
    elif gasto_total > 500:
        return "puede volver"
    else:
        return "✖️ definitivamente no va a volver"
    
print(prdecir_cliente(17, 1, 200.99))
print(prdecir_cliente(29, 8, 1700.99))
print(prdecir_cliente(328, 4, 65.28))
```

```python
from sklearn.ensemble import RandomForestClassifier
import numpy as np 

X_train = np.array([
    [10, 8, 450],
    [5, 12, 850],
    [20, 6, 50],
    [95, 1, 150],
    [120, 2, 50],
    [60, 8, 280],
    [15, 5, 350],
    [80, 8, 100]   
])

y_train = np.array([1, 1, 1, 0, 0, 0, 1, 0])

modelo = RandomForestClassifier(n_estimators=10,random_state=42)

modelo.fit(X_train, y_train)

clientes_nuevos = np.array([
    [17, 1, 200.99],
    [29, 8, 1700.99],
    [328, 4, 65.28]
])

predicciones = modelo.predict(clientes_nuevos)

for i, predicion in enumerate(predicciones):
    estado = "Va a volver " if predicion == 1 else "No va a volver"
    print(f"{[i]}: {estado}")
```
