### 2.2 — Limpieza de datos

🔧 **Ejercicio 2.2** — El dataset tiene algunos valores nulos (¡datos reales siempre tienen este problema!).  
1. Rellena los valores nulos de columnas **numéricas** con la **media** de cada columna
2. Verifica que ya no queden nulos
3. Explica en una celda de texto: ¿Por qué usamos la media y no simplemente eliminamos esas filas?

````python
# Tu código aquí

mediaEsperanza = df["esperanza_vida"].median()
mediaGasto = df["gasto_salud_usd"].median()
mediaMortalidad = df["mortalidad_infantil"].median()
mediaMedicos = df["medicos_por_1000"].median()
mediaObesidad = df["obesidad_pct"].median()
mediaDiabetes = df["diabetes_pct"].median()
mediaTabaquismo = df["tabaquismo_pct"].median()
mediaCamas = df["camas_hospital_por_1000"].median()
mediaVacunas = df["vacunacion_pct"].median()

df["esperanza_vida"]=df["esperanza_vida"].fillna(mediaEsperanza)
df["gasto_salud_usd"]=df["gasto_salud_usd"].fillna(mediaGasto)
df["mortalidad_infantil"]=df["mortalidad_infantil"].fillna(mediaMortalidad)
df["medicos_por_1000"]=df["medicos_por_1000"].fillna(mediaMedicos)
df["obesidad_pct"]=df["obesidad_pct"].fillna(mediaObesidad)
df["diabetes_pct"]=df["diabetes_pct"].fillna(mediaDiabetes)
df["tabaquismo_pct"]=df["tabaquismo_pct"].fillna(mediaTabaquismo)
df["camas_hospital_por_1000"]=df["camas_hospital_por_1000"].fillna(mediaCamas)
df["vacunacion_pct"]=df["vacunacion_pct"].fillna(mediaVacunas)
df.isnull().sum()
````

### 2.5 — Agrupaciones

🔧 **Ejercicio 2.5** — Usa `groupby()` para:
1. Calcular la **esperanza de vida promedio** por `region`
2. Calcular el **gasto en salud promedio** por `nivel_ingresos`
3. Calcular el **número de países** por región
4. ¿Qué región tiene la mayor esperanza de vida promedio? ¿Y la menor?

````python
# Tu código aquí
vida_region = df.groupby("region")["esperanza_vida"].mean()

print(vida_region)
gasto_ingresos = df.groupby("nivel_ingresos")["gasto_salud_usd"].mean()

print(gasto_ingresos)
paises_region = df.groupby("region")["pais"].count()

print(paises_region)
mayor = vida_region.idxmax()
menor = vida_region.idxmin()

print("Mayor esperanza de vida:", mayor)
print("Menor esperanza de vida:", menor)
````
